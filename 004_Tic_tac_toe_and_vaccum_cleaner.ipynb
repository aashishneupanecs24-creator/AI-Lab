```python
import random
import time

def initialize_rooms(num_rooms):
    """Initializes rooms with status provided by the user."""
    rooms = []
    print("\nEnter status for each room (0 for Clean, 1 for Dirty):")
    for i in range(num_rooms):
        while True:
            try:
                status = int(input(f"  Room {i+1} status: "))
                if status in [0, 1]:
                    rooms.append(status)
                    break
                else:
                    print("    Invalid input. Please enter 0 or 1.")
            except ValueError:
                print("    Invalid input. Please enter a number.")
    return rooms

def display_rooms(rooms):
    """Displays the current status of all rooms."""
    print("Current Room Status:")
    for i, status in enumerate(rooms):
        print(f"  Room {i+1}: {'Dirty' if status == 1 else 'Clean'}")

def vacuum_cleaner_agent(rooms):
    """Simulates a vacuum cleaner agent cleaning rooms."""
    print("\n--- Vacuum Cleaner Agent Started ---")
    display_rooms(rooms)

    cleaned_rooms_count = 0
    for i in range(len(rooms)):
        print(f"\nChecking Room {i+1}...")
        time.sleep(1) # Simulate agent movement/checking

        if rooms[i] == 1: # If room is dirty
            print(f"  Room {i+1} is Dirty. Cleaning...")
            time.sleep(1) # Simulate cleaning
            rooms[i] = 0 # Mark room as clean
            cleaned_rooms_count += 1
            print(f"  Room {i+1} is now Clean.")
        else:
            print(f"  Room {i+1} is Clean. Moving to next room.")

    print("\n--- Vacuum Cleaner Agent Finished ---")
    display_rooms(rooms)
    print(f"\nTotal rooms cleaned: {cleaned_rooms_count}")

    # Verify all rooms are clean
    if all(status == 0 for status in rooms):
        print("All rooms are successfully cleaned!")
    else:
        print("Warning: Some rooms might still be dirty (this shouldn't happen with this logic).")

# --- Demo ---
num_rooms = 5
initial_rooms = initialize_rooms(num_rooms)
vacuum_cleaner_agent(initial_rooms)
```

```python
def display_board(board):
    print("\n\t\t" + board[0] + " | " + board[1] + " | " + board[2])
    print("\t\t" + "-----------")
    print("\t\t" + board[3] + " | " + board[4] + " | " + board[5])
    print("\t\t" + "-----------")
    print("\t\t" + board[6] + " | " + board[7] + " | " + board[8] + "\n")

def check_win(board, player):
    # Check rows
    for i in range(0, 9, 3):
        if board[i] == board[i+1] == board[i+2] == player:
            return True
    # Check columns
    for i in range(3):
        if board[i] == board[i+3] == board[i+6] == player:
            return True
    # Check diagonals
    if board[0] == board[4] == board[8] == player or \
       board[2] == board[4] == board[6] == player:
        return True
    return False

def check_draw(board):
    for spot in board:
        if spot not in ['X', 'O']:
            return False
    return True

def tic_tac_toe():
    board = [str(i) for i in range(9)]  # Initialize board with numbers 0-8
    players = ['X', 'O']
    current_player_index = 0
    game_over = False

    print("Welcome to Tic-Tac-Toe!")
    print("Player 1: X, Player 2: O")
    print("\nUse numbers 0-8 to choose your spot on the board.\n")

    while not game_over:
        display_board(board)
        player = players[current_player_index]
        print(f"Player {player}'s turn.")

        while True:
            try:
                choice = int(input("Enter a number (0-8): "))
                if 0 <= choice <= 8 and board[choice] == str(choice):
                    break
                else:
                    print("Invalid move. Please choose an empty spot (0-8).")
            except ValueError:
                print("Invalid input. Please enter a number.")

        board[choice] = player

        if check_win(board, player):
            display_board(board)
            print(f"Congratulations! Player {player} wins!")
            game_over = True
        elif check_draw(board):
            display_board(board)
            print("It's a draw!")
            game_over = True
        else:
            current_player_index = (current_player_index + 1) % 2

    print("Game Over.")

tic_tac_toe()
```
