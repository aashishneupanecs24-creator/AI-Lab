```python
import random

def calculate_cost(state):
    n = len(state)
    cost = 0

    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def hill_climbing_with_restarts(n, initial_state):
    state = initial_state[:]
    restart_count = 0
    iteration = 0

    while True:
        current_cost = calculate_cost(state)

        print(f"\n[Restart {restart_count}, Iteration {iteration}] Current state: {state} | Cost: {current_cost}")

        if current_cost == 0:
            print(f"\nSolution found after {restart_count} restart(s)!")
            print("Final state:", state)
            return state

        best_state = state[:]
        best_cost = current_cost

        # Generate neighbors by swapping two queens
        for i in range(n):
            for j in range(i + 1, n):
                neighbor = state[:]
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

                neighbor_cost = calculate_cost(neighbor)

                if neighbor_cost < best_cost:
                    best_cost = neighbor_cost
                    best_state = neighbor

        # random restart
        if best_cost >= current_cost:
            print("Local minimum reached. Restarting with a random state...")
            state = list(range(n))
            random.shuffle(state)
            restart_count += 1
            iteration = 0
        else:
            state = best_state
            iteration += 1
```

```python
n = int(input("Enter N: "))
state = list(map(int, input("Enter initial row positions: ").split()))

if n < 1 or len(state) != n:
    print("Invalid input!")
elif sorted(state) != list(range(n)):
    print("Enter a permutation of rows from 0 to N-1.")
else:
  hill_climbing_with_restarts(n, state)
```
