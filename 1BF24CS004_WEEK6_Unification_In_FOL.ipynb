```python
import ast

# --- Unification Logic Functions ---
def is_variable(expr):
    return isinstance(expr, str) and expr[0].islower()

def is_compound(expr):
    return isinstance(expr, tuple) and len(expr) > 0

def occur_check(var, expr, subst):
    if var == expr:
        return True
    elif is_variable(expr) and expr in subst:
        return occur_check(var, subst[expr], subst)
    elif is_compound(expr):
        return any(occur_check(var, arg, subst) for arg in expr[1:])
    return False

def unify(x, y, subst=None):
    if subst is None:
        subst = {}
    if subst is False:
        return False
    while is_variable(x) and x in subst:
        x = subst[x]
    while is_variable(y) and y in subst:
        y = subst[y]
    if x == y:
        return subst
    if is_variable(x):
        return unify_variable(x, y, subst)
    if is_variable(y):
        return unify_variable(y, x, subst)
    if is_compound(x) and is_compound(y):
        if len(x) != len(y) or x[0] != y[0]:
            return False
        for arg_x, arg_y in zip(x[1:], y[1:]):
            subst = unify(arg_x, arg_y, subst)
            if subst is False:
                return False
        return subst
    return False

def unify_variable(var, val, subst):
    if occur_check(var, val, subst):
        return False
    new_subst = subst.copy()
    new_subst[var] = val
    return new_subst

def apply_substitution(expr, subst):
    """Recursively applies substitution dictionary to an expression."""
    if is_variable(expr):
        if expr in subst:
            return apply_substitution(subst[expr], subst)
        return expr
    elif is_compound(expr):
        return tuple(apply_substitution(arg, subst) for arg in expr)
    return expr

# --- Interactive Input Interface ---
print("=== Interactive FOL Unifier ===")
print("Format example for inputs:")
print("Variables (lowercase): 'x', 'y'")
print("Constants/Functions (uppercase): 'A', ('F', 'x')")
print("Complex expressions: ('P', 'x', 'B')\n")

try:
    # Get raw string inputs
    input1 = input("Enter Expression 1 (e.g., ('P', 'x', 'B')): ")
    input2 = input("Enter Expression 2 (e.g., ('P', 'A', 'y')): ")

    # Parse strings into Python tuples/objects safely
    expr1 = ast.literal_eval(input1)
    expr2 = ast.literal_eval(input2)

    print(f"\nUnifying: {expr1} and {expr2}")
    result = unify(expr1, expr2)

    if result is False:
        print("Result: Unification Failed!")
    else:
        print(f"Substitution (MGU): {result}")
        unified_expr = apply_substitution(expr1, result)
        print(f"Final Unified Form: {unified_expr}")
except Exception as e:
    print(f"Error parsing inputs: {e}. Please ensure inputs are in valid Python tuple/string format.")
```
