```python
# Forward Chaining in First Order Logic

# Step 1: Initialize Knowledge Base with facts
KB = [
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
]

# Step 2: Define the goal
GOAL = ("Criminal", "Robert")

print("Initial Knowledge Base:")
for fact in KB:
    print(fact)

# Step 3: Forward Chaining
while True:

    NEW_FACTS = []

    # Rule 1: Missile(x) -> Weapon(x)
    for fact in KB:
        if fact[0] == "Missile":
            new_fact = ("Weapon", fact[1])
            if new_fact not in KB and new_fact not in NEW_FACTS:
                NEW_FACTS.append(new_fact)

    # Rule 2: Missile(x) AND Owns(A,x)
    # -> Sells(Robert,x,A)
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]

            if ("Owns", "A", x) in KB:
                new_fact = ("Sells", "Robert", x, "A")

                if new_fact not in KB and new_fact not in NEW_FACTS:
                    NEW_FACTS.append(new_fact)

    # Rule 3: Enemy(x,America) -> Hostile(x)
    for fact in KB:
        if fact[0] == "Enemy" and fact[2] == "America":
            new_fact = ("Hostile", fact[1])

            if new_fact not in KB and new_fact not in NEW_FACTS:
                NEW_FACTS.append(new_fact)

    # Rule 4: American(p) AND Weapon(q)
    # AND Sells(p,q,r) AND Hostile(r)
    # -> Criminal(p)

    for fact in KB:
        if fact[0] == "American":
            p = fact[1]

            for weapon in KB:
                if weapon[0] == "Weapon":
                    q = weapon[1]

                    for sale in KB:
                        if sale[0] == "Sells" and sale[1] == p and sale[2] == q:
                            r = sale[3]

                            if ("Hostile", r) in KB:
                                new_fact = ("Criminal", p)

                                if new_fact not in KB and new_fact not in NEW_FACTS:
                                    NEW_FACTS.append(new_fact)

    # Step 4: Check whether goal is generated
    print("\nNew Facts Generated:")
    for fact in NEW_FACTS:
        print(fact)

    if GOAL in NEW_FACTS or GOAL in KB:
        print("\nGOAL PROVED!")
        print("Robert is a Criminal.")
        break

    # Step 5: If no new facts, stop
    if not NEW_FACTS:
        print("\nGOAL CANNOT BE PROVED!")
        break

    # Step 6: Update Knowledge Base
    KB.extend(NEW_FACTS)
```
