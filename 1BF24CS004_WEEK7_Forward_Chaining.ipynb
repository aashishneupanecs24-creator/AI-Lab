```python
# Forward Chaining in First Order Logic with Intermediate Steps

# Step 1: Initialize Knowledge Base with facts
KB = [
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
]

# Step 2: Define the goal
GOAL = ("Criminal", "Robert")

print("=== STARTING STATE ===")
print("Initial Knowledge Base:")
for fact in KB:
    print(f"  - {fact}")

step_counter = 1

# Step 3: Forward Chaining Loop
while True:
    print(f"\n=== ITERATION {step_counter} ===")
    NEW_FACTS = []

    # Rule 1: Missile(x) -> Weapon(x)
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]
            new_fact = ("Weapon", x)
            if new_fact not in KB and new_fact not in NEW_FACTS:
                print(f"[Rule 1 Triggered] Missile({x}) matches fact {fact}.")
                print(f"  -> Derived: Weapon({x})")
                NEW_FACTS.append(new_fact)

    # Rule 2: Missile(x) AND Owns(A,x) -> Sells(Robert,x,A)
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]
            if ("Owns", "A", x) in KB:
                new_fact = ("Sells", "Robert", x, "A")
                if new_fact not in KB and new_fact not in NEW_FACTS:
                    print(f"[Rule 2 Triggered] Missile({x}) and Owns(A, {x}) match.")
                    print(f"  -> Derived: Sells(Robert, {x}, A)")
                    NEW_FACTS.append(new_fact)

    # Rule 3: Enemy(x, America) -> Hostile(x)
    for fact in KB:
        if fact[0] == "Enemy" and fact[2] == "America":
            x = fact[1]
            new_fact = ("Hostile", x)
            if new_fact not in KB and new_fact not in NEW_FACTS:
                print(f"[Rule 3 Triggered] Enemy({x}, America) matches fact {fact}.")
                print(f"  -> Derived: Hostile({x})")
                NEW_FACTS.append(new_fact)

    # Rule 4: American(p) AND Weapon(q) AND Sells(p,q,r) AND Hostile(r) -> Criminal(p)
    for fact in KB:
        if fact[0] == "American":
            p = fact[1]
            for weapon in KB:
                if weapon[0] == "Weapon":
                    q = weapon[1]
                    for sale in KB:
                        if sale[0] == "Sells" and sale[1] == p and sale[2] == q:
                            r = sale[3]
                            if ("Hostile", r) in KB:
                                new_fact = ("Criminal", p)
                                if new_fact not in KB and new_fact not in NEW_FACTS:
                                    print(f"[Rule 4 Triggered] American({p}), Weapon({q}), Sells({p}, {q}, {r}), and Hostile({r}) match.")
                                    print(f"  -> Derived: Criminal({p})")
                                    NEW_FACTS.append(new_fact)

    # Step 4: Show facts generated in this loop iteration
    print("\nSummary of New Facts this iteration:")
    if not NEW_FACTS:
        print("  None")
    else:
        for fact in NEW_FACTS:
            print(f"  + {fact}")

    # Step 5: Check whether goal is generated or exists
    if GOAL in NEW_FACTS or GOAL in KB:
        # Ensure it is in the KB for the final printout
        KB.extend(NEW_FACTS)
        print("\n=== GOAL PROVED! ===")
        print(f"The goal {GOAL} has been successfully proved.")
        print("Final Knowledge Base:")
        for fact in KB:
            print(f"  - {fact}")
        break

    # Step 6: If no new facts, stop to prevent infinite loop
    if not NEW_FACTS:
        print("\n=== GOAL CANNOT BE PROVED! ===")
        break

    # Step 7: Update Knowledge Base
    KB.extend(NEW_FACTS)
    step_counter += 1
```
